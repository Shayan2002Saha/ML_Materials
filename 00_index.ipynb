## 0. How to Use This MaterialThis chapter is the only part of the material that is not machine-learning content. It documents theconventions that keep the structure stable while the chapters are filled in.

### 0.1 Purpose and ScopeA single, complete classical-ML reference — deep enough to learn from the first time, and structured enough tolook something up in later.* The chapter list (top level) is meant to stay **stable**.* Sub-sections are meant to be **split, renamed and deepened** as chapters are written.* Nothing here is a syllabus to be raced through — it is a set of containers to be filled properly.

### 0.2 What This Material Covers, and What It Does Not| In scope | Out of scope ||---|---|| Supervised learning on tabular data — regression and classification | Neural networks, backpropagation, CNNs, RNNs, transformers || Unsupervised learning — clustering, anomaly detection, association rules | Computer vision and audio || The full data pipeline — EDA, preprocessing, feature engineering, selection | Large-scale NLP and language models || Evaluation, validation, tuning, interpretability, deployment basics | Reinforcement learning beyond a one-paragraph definition (1.3) || The mathematics each algorithm actually needs (Chapter 2) | Distributed / big-data training (Spark, Dask) |Where a deep-learning method is the honest answer to a problem, the material says so and points at it ratherthan pretending a classical model is better (1.7).

### 0.3 Per-Concept Documentation TemplateFull style guide: [`CLAUDE.md`](./CLAUDE.md). Summary — every concept is **explanation + runnable demo, neverone alone**:1. A markdown cell with a real explanation — not a one-liner. Use whatever fits best: a small table (metrics,   kernels, encoders, linkage criteria), a short "why it matters" note, a named callout for a nuance or gotcha   (`**Common mistake** — ...`), or a contrast with a related/confusable concept.2. One or more **actual code cells** (not fenced ` ```python ` blocks inside markdown), each minimal and   focused on one thing, with an inline `#` comment on the lines whose output matters.3. A concept-specific gotcha or a second angle worth showing gets its own small markdown note + its own small   code cell, rather than one giant cell for everything.Make abstract ideas tangible. For every algorithm the order is fixed:```textthe objective, written out  ->  a from-scratch NumPy implementation  ->  the scikit-learn equivalent                            ->  an assertion that the two agree```The from-scratch version is what makes the mechanism visible; the scikit-learn version is what gets usedeverywhere afterwards. Neither one alone is enough.

### 0.4 Cell Rhythm for Each ConceptThe cell sequence when a concept is written up:```textMarkdown  ->  explanation (table / prose / "why it matters")Code      ->  minimal runnable demo, inline # comments on the lines that matterMarkdown  ->  (optional) a named gotcha or second angleCode      ->  (optional) its own small demo cell```Repeat per concept/sub-concept. Each chapter ends with a scratch cell for ad-hoc experiments — promoteanything worth keeping into a proper section instead of leaving it there.

### 0.5 Numbering and Insertion RulesConventions that avoid renumbering everything every time something is added:1. **Chapter numbers are frozen.** New major domains are appended, never inserted.2. **New sub-sections are appended** to the end of a chapter, before the `x.9` bucket (if the chapter has one   — see rule 3).3. **`x.9`** is an *optional* per-chapter overflow bucket. Anything with no obvious home can land here first   and migrate once a pattern emerges — a chapter that is otherwise complete does not need to keep an empty   one around.4. **Deeper levels are free-form.** `17.3.2`, `17.3.3` etc. can be created and reordered without touching   anything above them.5. **Moving, renaming or removing a section** means fixing every pointer that referenced the old location — in   other chapters and in the Appendix (§25.1, §25.2) — rather than leaving it dangling.

### 0.6 Running the NotebooksDependencies are managed with [uv](https://docs.astral.sh/uv/); `pyproject.toml` and `uv.lock` are committed,so the environment is reproducible exactly.```bashuv sync                      # create .venv and install the locked dependency setuv run jupyter lab           # open the notebooks```To re-execute a chapter from a terminal:```bashuv run jupyter nbconvert --to notebook --execute --inplace 11_linear_regression.ipynb```Every notebook in this repository is committed **with its outputs**, so plots and tables render on GitHubwithout running anything. Every stochastic call is seeded (0.7), so re-executing a chapter reproduces thecommitted output rather than a slightly different one.

### 0.7 Datasets and DeterminismNo notebook downloads anything. Data comes from exactly three places:| Source | Used for | Examples ||---|---|---|| `sklearn.datasets` bundled loaders | Realistic small problems | `load_iris`, `load_wine`, `load_breast_cancer`, `load_diabetes`, `load_digits` || `sklearn.datasets` generators | Controlled shapes where the right answer is known | `make_classification`, `make_regression`, `make_blobs`, `make_moons`, `make_circles` || [`data/`](./data) in this repository | Messy realistic tabular data for Chapters 4–6 | `customers.csv`, `transactions.csv` |**Determinism.** Every split, generator, resampler, initializer and optimizer that can take a seed is givenone — the constant `RANDOM_STATE = 42` is defined at the top of each chapter that needs it. Two consequences:a committed output can be trusted as the output the code actually produced, and any difference on re-executionis a real change, not noise.

### 0.9 Reading OrderThe chapters are written to be read in order — each one assumes the vocabulary of the ones before it. Chapters 2 and 3 exist to be  referred back to, not necessarily read first.* **Only want a specific algorithm?** Chapters 11–21 are individually readable, but each one reports its  results using Chapter 8's metrics and reasons about fit using Chapter 9's vocabulary. ### 25.3 NotationThe symbols used consistently across the algorithm chapters.| Symbol | Meaning ||---|---|| $n$ | number of samples (rows) || $p$ | number of features (columns) || $X$ | feature matrix, shape $(n, p)$ || $\mathbf{x}^{(i)}$ | the $i$-th sample — one row of $X$ || $x_j$ | the $j$-th feature — one column of $X$ || $y$ | target vector, shape $(n,)$; $y^{(i)}$ is one sample's target || $\hat{y}$ | predicted target || $\mathbf{w}$, $b$ | learned weights (coefficients) and intercept (bias) || $\theta$ | all learned parameters of a model, taken together || $\mathcal{L}$ | loss on a single sample || $J(\theta)$ | cost — the loss averaged or summed over the whole training set || $\alpha$ | learning rate (Chapter 10) or regularization strength (Chapter 12) — the two standard uses collide, so each chapter says which it means || $\lambda$ | regularization strength in the mathematics; scikit-learn spells it `alpha` (12.1) || $K$ | number of clusters (Chapter 20) or number of CV folds (8.3) || $k$ | number of neighbors (Chapter 14) |Vectors are column vectors in the mathematics and 1-D NumPy arrays in the code; the material flags the placeswhere that distinction actually changes a result.
